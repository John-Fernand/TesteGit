```python
import numpy as np

np.set_printoptions(suppress=True)

dados = np.array([
                   [120, 15000, 100, 125.00, 4],
                   [135, 16200, 108, 120.00, 5],
                   [98, 11700, 82, 119.39, 3],
                   [150, 21000, 125, 140.00, 7],
                   [175, 24500, 140, 140.00, 9],
                   [160, 22400, 132, 140.00, 6],
                   [110, 13200, 95, 120.00, 2]
                  ])
dias = np.array([
"Segunda",
"Terça",
"Quarta",
"Quinta",
"Sexta",
"Sábado",
"Domingo"
])
print(dados)
print(dias)
```

```python
'''Parte 1 — Conhecendo os dados
1.1 - Mostre:

    quantidade de linhas;
    quantidade de colunas;
    quantidade total de valores;
    número de dimensões do array;
    tipo dos dados armazenados.

Utilize:.shape
.size
.ndim
.dtype'''

tamanho = np.shape(dados) #Estou utilizando a função do numpy shape para receber uma tupla com o número de linhas e colunas do array. Estou armazenando esses valores em tamanho.
print(f"O número de linhas desse array é {tamanho[0]}\nO número de colunas desse array é de {tamanho[1]}.")
print(f"O número total de elementos desse aray é de {tamanho[0]*tamanho[1]}.") #Utilizei o objeto tamanho para calcular, porém é possível receber o número de elementos utilizando o np.size.
#print(f"O número total de elementos desse aray é de {np.size(dados)}.") Esse é o jeito utilizando o np.size.
print(f"O número de dimensões do array é de {np.ndim(dados)}.")
#print(f"O número de dimensões do array é de {len(tamanho)}.") Jeito alternativo de fazer utilizando o tamanho o número de elementos de np.shape, os quais retornam o número de dimensões.
print(f"O tipo dos dados armazenados é {dados.dtype}.")
```

```python
'''
Parte 2 — Acessando os dados
2.1 - Mostre os dados referentes à sexta-feira.
2.2 - Mostre somente o faturamento de todos os dias.
2.3 - Mostre somente a quantidade de reclamações.
2.4 - Mostre os dados de quinta-feira até domingo.
2.5 - Mostre somente as colunas:

    pedidos;
    faturamento;
    clientes.'''
print(f"Os dados referentes à sexta-feira são: {dados[4]}.") #Mostrei apenas os dados da quinta linha, ou seja, sexta-feira.
print(f"Os faturamentos de todos os dias são: {dados[:, 1]}.")#Utilizo o : para obter todos os valores de uma linha e utilizo o 4 para determinar de qual coluna, ou seja, obtenho os valores apenas das linhas de uma coluna especifica.
um = np.ones((1, len(dados[:, 4])))
total = um @ dados[:, 4].reshape(-1, 1)
total = total.item(0)
print(f"O número de reclamações de todos os dias são: {dados[:, 4]}. \nO número total de reclamações é {total}. ") #Utilizei a multiplicação de matrizes para obter uma matriz 1x1 com o resultado das multiplicações por conta das limitações do uso do sum e for. OBS: Acabei por utilizar IA para resolver certos poblemas de sintaxe, a ideia de multiplicação veio de mim.
print(f"Os dados de quinta até domingo é de {dados[3:, :]}")
print(f"Pedidos: {dados[:, 0]}.\nFaturamento: {dados[:,1]}.\nClientes: {dados[:,2]}.\nTicket_Medio: {dados[:,3]}.\nReclamações: {dados[:,4]}.  ")
```

```python
'''Parte 3 — Operações numéricas
3.1- Calcule o total de pedidos realizados durante a semana.
3.2 - Calcule o faturamento total.
3.3 - Calcule a quantidade total de clientes.
3.4 - Calcule a média diária de pedidos.
3.5 - Calcule a média diária de faturamento.
3.6 - Calcule a média de reclamações por dia.
utilize funções do NumPy, evitando utilizar for.'''
#Soluções realizadas com multiplicações de matrizes para transformar em uma matriz 1x1 que possui o valor da soma do array original.
#3.1
um = np.ones((1, len(dados[:, 0])))
total = um @ dados[:, 0].reshape(-1, 1)
total = total.item(0)
print(f"O número total de pedidos é de {total}.")

#3.2
um = np.ones((1, len(dados[:, 1])))
total = um @ dados[:, 1].reshape(-1, 1)
total = total.item(0)
print(f"O faturamento total foi de {total}")

#3.3
um = np.ones((1, len(dados[:, 2])))
total = um @ dados[:, 2].reshape(-1, 1)
total = total.item(0)
print(f"A quantidade total de clientes é de {total}")
#Questões seguintes utilizam o valor das somas obtidas pelo método de multiplicação de matrizes e dividindo esse resultado pelo número de itens da coluna.
#3.4
um = np.ones((1, len(dados[:, 0])))
total = um @ dados[:, 0].reshape(-1, 1)
total = total.item(0)
total = total/len(dados[:, 0])
print(f"A média diária de pedidos é de aproximadamente {round(total,2)}.")#Arredondei os resultados para ficarem mais claros.

#3.5
um = np.ones((1, len(dados[:, 1])))
total = um @ dados[:, 1].reshape(-1, 1)
total = total.item(0)
total = total/len(dados[:,1])
print(f"A média diária de faturamento é de aproximadamente {round(total,2)}.")

#3.6
um = np.ones((1, len(dados[:, 4])))
total = um @ dados[:, 4].reshape(-1, 1)
total = total.item(0)
total = total/len(dados[:,4])
print(f"A média total de reclamações é de aproximadamente {round(total,2)}.")

```

```python
'''Parte 4 — Análise por dia
Agora precisamos descobrir quais foram os melhores e piores resultados.4.1 - Identifique o dia com:

    maior número de pedidos;
    menor número de pedidos.

Utilize:np.argmax()
np.argmin()


Depois utilize o resultado para descobrir o nome do dia.4.2 - Faça o mesmo para o faturamento.
4.3 - Calcule a amplitude do faturamento:

    maior faturamento
    menor faturamento'''
#4.1
#Toda a ideia inicial seria criar uma lista de dicionários com as chaves id e dia.
#Observei que provavelmente teria que utilizar for se fizesse desse jeito, então
#acabei optando por remover a chave id. Utilizei apenas o índice.
#O o cálculo obtem um valor, esse valor indica um dia da semana pelo indice da lista.
#Tudo isso foi em vão pois era só utilizar o array inicial de dias ao invés de fazer
#uma nova lista.
lista = []
lista.extend([{"dia": "segunda"}, {"dia": "terça"}, {"dia": "quarta"}, {"dia": "quinta"}, {"dia": "sexta"}, {"dia": "sábado"}, {"dia": "domingo"} ])
maiorpedidos = np.argmax(dados[:,0])
menorpedidos = np.argmin(dados[:,0])
print(f"O dia com o maior número de pedidos foi {lista[maiorpedidos]["dia"]}.\nO dia com o menor número de pedidos foi {lista[menorpedidos]["dia"]}.")
#4.2
maiorfaturamento = np.argmax(dados[:,1])
menorfaturamento = np.argmin(dados[:,1])
print(f"O dia com o maior faturamento foi {lista[maiorpedidos]["dia"]}.\nO dia com o menor faturamento foi {lista[menorpedidos]["dia"]}.")
#4.3
amplitude = (dados.item(maiorfaturamento, 1) - dados.item(menorfaturamento, 1))
print(f"A amplitude do faturamento é de {amplitude}.")
```

```python
'''Parte 5 — Máscaras e filtros
Agora queremos identificar situações específicas.
5.1 - Mostre os dias em que o faturamento foi superior a 12000.
5.2 - Mostre os dias em que foram realizados mais de 150 pedidos.
5.3 - Mostre os dias que tiveram mais de 4 reclamações.
5.4 - Mostre os faturamentos dos dias que tiveram mais de 150 pedidos.
5.5 - Quantos dias tiveram faturamento abaixo da média semanal?'''
#5.1
mascara = (dados[:,1] > 12000)
print(dias[mascara])

#5.2
mascara = (dados[:, 0] > 150)
print(dias[mascara])

#5.3
mascara = (dados[:, 4] > 4)
print(dias[mascara])

#5.4
mascara = (dados[:, 0] > 150)
for i in (dados:[:, 1]):
  print()
print(dias[mascara])

#5.5
mascara = (dados[:, 0] > 150)
print(dias[mascara])
```

```python
#Parte 6 — Classificação dos dias
classificação = np.where(
dados[:, 1]>=20000,
"Alto desempenho",
"Desempenho normal"
)
classificaçãoalta = 0
classificaçãonormal = 0
for i in range(len(classificação)):
  print(f"A classificação de {dias[i]} foi de {classificação[i]}. ")
  if classificação[i] == "Alto desempenho":
    classificaçãoalta += 1
  else:
    classificaçãonormal += 1
print(f"Quantidade de dias com classificação alta = {classificaçãoalta} dias.\nQuantidade de dias com classificação normal = {classificaçãonormal} dias.")

```

```python
#Parte 7 — Estatística
print(f"A média do faturamento foi de {round(np.average(dados[:,1]), 2)}). ")
print(f"A mediana do faturamento foi de {np.median(dados[:,1])}.")
print(f"O desvio padrão do faturamento foi de {round(np.std(dados[:,1]),2)}.")
print(f"O faturamento mínimo foi de {round( np.min(dados[:,1]),2)}. ")
print(f"O faturamento máximo foi de {round( np.max(dados[:,1]),2)}.")
print(f"O faturamento teve uma grande variação, como indicado de acordo com o desvio padrão.")
```

```python
'''Parte 8 — Trabalhando com axis
Agora queremos analisar os dados de outra forma.
Calcule:8.1 - A média de cada indicador durante a semana.
Utilize:            np.mean(dados, axis=0) '''
lista = ()
np.set_printoptions(suppress=True)
medias = np.round(np.mean(dados, axis=0), 2)
print(medias)
print(f"O maior valor observado em cada indicador é {np.max(dados, axis=0)}.")
print(f"O menor valor observado em cada indicador é {np.min(dados, axis=0)}.")
variação = np.std(dados, axis=0)
coeficiente = (variação / medias) * 100
print(f"A maior variação relativa {medias[np.argmax(coeficiente)]}.")
```
