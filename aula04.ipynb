```python
# Usamos o comando return para retornar valores de uma função e encerrar sua execução
# Função para calcular percentual de um valor qualquer
# É possível deixar um valor fixo para um argumento,
# quando ele não é informado na chamada, o valor padrão ´e utilizado
def calcula_percento(valor, percentual=100):
  quanto = valor * percentual / 100
  return quanto

print ('20% de 300 são: {:10.2f}'.format (calcula_percento (300,20)))

print ('100% de 50 são: {:10.2f}'.format (calcula_percento (50)))
```

```python
primo = True
número = int(input("Digite um número"))
if número == 1 or número == 0:
  primo = False
if número % 2 == 0 and número != 2:
  primo = False
else:
  for i in range(3, número, 2):
    if número % i == 0:
      primo = False
if primo == True:
  print("O número {} é primo".format(número))
if primo == False:
  print("O número {} não é primo".format(número))
```
