```python
# Uso do comando while para repetição
senha = "112233"
digitado =" "
#Vai repetir o recho de código enquanto a leitura for diferente da senha
#Leitura: enquanto digitado é diferente de senha, repita
while (digitado != senha):
  digitado = input("Digite a senha: ")
  if digitado == senha :
    print('Acesso liberado ')
  else:
    print('Senha incoreta. Tente novamente')
```

```python
# For para passar pela lista
# Lista de notas de provas
notas_provas = [9, 6.5, 3.4, 8.5]
soma = 0
for nota in notas_provas:
  soma = soma + nota

média = soma / 4
# sem saber a quantidade de notas
# len(notas_provas)
print('Média =', média)
```
