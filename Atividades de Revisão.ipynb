Escreva um programa em Python que gere um número aleatório de 0 a 15, usando a biblioteca 'random' e o método 'randint()'. Na sequência, utilizando um loop faça com que o usuário tente acertar o número gerado.
A cada vez que o número é acertado, armazene num arquivo o número gerado e a quantidade de tentativas que o usuário realizou para acertar. Quando o usuário decidir parar de adivinhar, mostre todo o histórico que ele construiu durante a execução do programa.

Atividade II

```python
import random
import pandas

listadicionario = []
repetir = True
num_escolhido = -1
tentar = input("Gostaria de continuar tentando? Digite stop caso queira parar. ")
if tentar == ("stop"):
  repetir = False
while repetir == True:
  tentativas = 0
  num_al = random.randint(0, 15)
  while num_escolhido != num_al:
    try:
      num_escolhido = int(input("Insira um número. "))
      if num_escolhido == num_al:
        print("Número correto")
        tentativas += 1
      else:
        print("Número incorreto, tente novamente. ")
        tentativas += 1
    except ValueError:
      print("Valor inválido, insira um número inteiro. ")
      tentativas += 1
  dicionario = {'Número': num_escolhido, 'Número de Tentativas': tentativas}
  listadicionario.append(dicionario)
  df = pandas.DataFrame(listadicionario)
  df.to_csv("dados.csv", index=False)
  tentar = input("Gostaria de continuar tentando? Digite stop caso queira parar. ")
  if tentar == ("stop"):
    repetir = False
df = pandas.read_csv("dados.csv")
print(df)
```
