Fonte de dados: https://store.steampowered.com/


```python
import requests #Requisições para obter o conteúdo HTML das páginas.
import pandas as pd #Manipulação e organização de dados em dataframes.
import time #Intervalo entre requisições.
from bs4 import BeautifulSoup #Analisar o HTML e localizar os elementos necessários.
from google.colab import files #Baixar o arquivo final.
```

```python
app_ids = [] #Lista que será utilizada para armazenar o AppId de cada jogo.
while True:
    app_id = input("Digite o AppID do jogo (digite 'sair' para finalizar): \n")
    if app_id == "sair": #Sai do loop quando o usuário quiser.
        break
    if app_id.isdigit(): #Verifica se é um digito, caso seja, adiciona o AppID para a lista, convertendo-o para um número inteiro.
        app_ids.append(int(app_id))
        print("AppID adicionado!")
    else:
        print("Valor inválido!") #Informa que o valor fornecido é inválido.
print("\nAppIDs selecionados:", app_ids)
```

```python
jogos = [] #Lista que será utilizada para armazenar os dados dos jogos.
for app_id in app_ids: #Percore cada AppID da lista.
    url = f"https://store.steampowered.com/app/{app_id}/?cc=br&l=brazilian" #Endereço que será utilizado.
    response = requests.get(url) #Requisição HTTP do conteúdo.
    soup = BeautifulSoup(response.text, "html.parser") #HTML retornado.
    nome = soup.select_one(".apphub_AppName") #Interpreta o HTML para localizar os elementos desejados.
    desenvolvedor = soup.select("#developers_list a")
    lancamento = soup.select_one(".release_date .date")
    preco = soup.select_one(".discount_final_price") #Caso o jogo esteja em promoção, será pego o valor com desconto.
    if not preco:
        preco = soup.select_one(".game_purchase_price") #Preço em desconto.
    desconto = soup.select_one(".discount_pct") #Percentual do desconto.
    preco = preco.get_text(strip=True) if preco else "Não disponível" #Verifica se há algum valor, caso contrário, substitui por "Não disponível".
    desconto = desconto.get_text(strip=True) if desconto else "Sem desconto" #Verifica se há algum valor, caso contrário, substitui por "Sem desconto".
    jogos.append({ #Armazena os dados.
        "AppID": app_id,
        "Nome": nome.text.strip() if nome else None,
        "Desenvolvedora": desenvolvedor[0].text.strip() if desenvolvedor else None,
        "Lancamento": lancamento.text.strip() if lancamento else None,
        "Preco": preco,
        "Desconto": desconto
    })
    time.sleep(1) #Intervalo de 1 segundo.
df = pd.DataFrame(jogos) #Transforma a lista de dicionários em um dataframe.
print(df)
```

```python
df.to_csv("steam_games.csv", index=False, encoding="utf-8-sig") #Exporta o dataframe para um arquivo CSV.
```
