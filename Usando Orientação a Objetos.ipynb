```python
class Calculadora:
  def adição(self):
    a = float(input("Selecione um valor para a "))
    b = float(input("Selecione um valor para b "))
    resultado = a + b
    print(f"O resultado da soma de {a} + {b} é de {resultado}")
  def subtração(self):
    a = float(input("Selecione um valor para a "))
    b = float(input("Selecione um valor para b "))
    resultado = a - b
    print(f"O resultado da subtração de {a} - {b} é de {resultado}")
  def multiplicação(self):
    a = float(input("Selecione um valor para a "))
    b = float(input("Selecione um valor para b "))
    resultado = a * b
    print(f"O resultado da multiplicação de {a} * {b} é de {resultado}")
  def divisão(self):
    a = float(input("Selecione um valor para a "))
    b = float(input("Selecione um valor para b "))
    resultado = a / b
    print(f"O resultado da divisão de {a} por {b} é de {resultado} ")
  def exponenciação(self):
    a = float(input("Selecione um valor para a "))
    b = float(input("Selecione um valor para b "))
    resultado = a ** b
    print(f"O resultado da exponenciação de {a} pelo expoente {b} é de {resultado}")
  def raiz_quadrada(self):
    a = float(input("Digite um número "))
    p = (a + (a/a) )/2.
    b = p
    for i in range(10):
        p = (b + (a/b) )/2.
        b = p
    print(f"A raiz quadrada de {a} é aproximadamente {p} ")
  def menu(self):
    print("Qual operação você deseja realizar? ")
    print("1. Adição. ")
    print("2. Subtração. ")
    print("3. Multiplicação. ")
    print("4. Divisão. ")
    print("5. Exponenciação. ")
    print("6. Radiciação. ")
    print("7. Finalizar. ")
  def executar(self):
    continuar = True
    while continuar == True:
      self.menu()
      escolha = input("Digite a opção. ")
      match escolha:
        case "1":
          self.adição()
        case "2":
          self.subtração()
        case "3":
          self.multiplicação()
        case "4":
          self.divisão()
        case "5":
          self.exponenciação()
        case "6":
          self.raiz_quadrada()
        case "7":
          continuar = False
          print("Processo encerrado. ")
        case _:
          print("Opção inválida. ")

calculadora = Calculadora()
calculadora.executar()



```
