```python
#Código corrigido com chatgpt por leves problemas com filtro.
class Weaponpicker:

  def escolhertipoarma(self, df, listatiposarmas):
    while True:
      try:
        print("\nSelecione o tipo de arma desejado.")

        for i, tipo in enumerate(listatiposarmas):
          print(f"{i} para escolher {tipo}.")

        tipodesejado = int(input("Digite o número do tipo de arma: "))

        if not 0 <= tipodesejado < len(listatiposarmas):
          raise ValueError

        return tipodesejado

      except ValueError:
        print("Valor inválido, tente novamente.")

  def requerimentostatus(self, df, statusds):
    while True:
      try:
        print("\nSelecione o atributo que deseja verificar.")

        for i, status in enumerate(statusds):
          print(f"{i} para escolher {status}.")

        requerimentodesejado = int(input("Digite o número do atributo: "))

        if not 0 <= requerimentodesejado < len(statusds):
          raise ValueError

        valorstatus = int(input("Digite o valor do seu status (1 a 99): "))

        if not 1 <= valorstatus <= 99:
          raise ValueError

        return requerimentodesejado, valorstatus

      except ValueError:
        print("Valor inválido. Tente novamente.")

  def scalingdesejado(self, df, valorscaling, scalingtipo):
    while True:
      try:
        print("\nSelecione o atributo do scaling.")

        for i, atributo in enumerate(scalingtipo):
          print(f"{i} para escolher {atributo}.")

        scalingtipodesejado = int(input("Digite o número do atributo: "))

        if not 0 <= scalingtipodesejado < len(scalingtipo):
          raise ValueError

        print("\nSelecione a letra de scaling desejada.")

        for i, letra in enumerate(valorscaling):
          print(f"{i} para escolher {letra}.")

        scalingdesejado = int(input("Digite o número da letra: "))

        if not 0 <= scalingdesejado < len(valorscaling):
          raise ValueError

        return scalingdesejado, scalingtipodesejado

      except ValueError:
        print("Valor inválido. Tente novamente.")

  def filtragem(
      self, df, listatiposarmas, escolhertipoarma2,
      statusds, requerimentodesejado2, valorstatus2,
      valorscaling, scalingtipo, scalingdesejado2,
      scalingtipodesejado2
  ):
    tipoarma = listatiposarmas[escolhertipoarma2]
    statusdesejado = statusds[requerimentodesejado2]
    coluna_scaling = scalingtipo[scalingtipodesejado2]
    valor_scaling = valorscaling[scalingdesejado2]

    # Normalizar texto para evitar diferenças de espaços e maiúsculas.
    tipos = df["Weapon Type"].astype("string").str.strip()
    escalas = (
        df[coluna_scaling]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    # Converter os requisitos para números.
    requisitos = pd.to_numeric(df[statusdesejado], errors="coerce")

    # Criar cada filtro separadamente.
    filtro_tipo = tipos.str.casefold() == tipoarma.casefold()
    filtro_status = requisitos <= valorstatus2
    filtro_scaling = escalas == valor_scaling

    # Diagnóstico.
    print("\n===== DIAGNÓSTICO =====")
    print(f"Tipo de arma: {tipoarma}")
    print(f"Requisito consultado: {statusdesejado}")
    print(f"Status informado: {valorstatus2}")
    print(f"Coluna de scaling: {coluna_scaling}")
    print(f"Scaling escolhido: {valor_scaling}")

    print(f"\nCorrespondências de tipo: {filtro_tipo.sum()}")
    print(f"Correspondências de requisito: {filtro_status.sum()}")
    print(f"Correspondências de scaling: {filtro_scaling.sum()}")

    filtro_completo = filtro_tipo & filtro_status & filtro_scaling

    print(f"Correspondências dos três filtros: {filtro_completo.sum()}")

    # Mostrar os resultados.
    df_filtro = df.loc[filtro_completo]

    print("\n===== ARMAS ENCONTRADAS =====")

    if df_filtro.empty:
      print("Nenhuma arma corresponde aos três critérios.")

      # Mostrar armas do tipo escolhido para ajudar na investigação.
      armas_tipo = df.loc[filtro_tipo]

      print("\nExemplos de armas desse tipo:")
      print(
          armas_tipo[
              ["Weapon", statusdesejado, coluna_scaling]
          ].head(20).to_string(index=False)
      )

    else:
      print(df_filtro.to_string(index=False))

    print(f"\nTotal de armas encontradas: {len(df_filtro)}")

    return df_filtro

  def menu(self):
    print("\n===== WEAPON PICKER =====")
    print("1. Escolher o tipo de arma.")
    print("2. Escolher o status do personagem.")
    print("3. Escolher o scaling desejado.")
    print("4. Executar a filtragem.")
    print("5. Encerrar o processo.")

  def executar(self, df):
    continuar = True

    escolhertipoarma2 = None
    requerimentodesejado2 = None
    valorstatus2 = None
    scalingdesejado2 = None
    scalingtipodesejado2 = None

    scalingtipo = (
        "Scaling (+10/+5) STR",
        "Scaling (+10/+5) DEX",
        "Scaling (+10/+5) INT",
        "Scaling (+10/+5) FTH"
    )

    statusds = (
        "Requirements STR",
        "Requirements DEX",
        "Requirements INT",
        "Requirements FTH"
    )

    valorscaling = ("E", "D", "C", "B", "A", "S")

    listatiposarmas = (
        "Axe", "Bow", "Claw", "Crossbow",
        "Curved Greatsword", "Curved Sword", "Dagger", "Fist",
        "Great Hammer", "Greataxe", "Greatbow", "Greatsword",
        "Halberd", "Hammer", "Katana", "Piercing Sword",
        "Reaper", "Sacred Chime", "Spear", "Staff",
        "Straight Sword", "Talisman", "Ultra Greatsword", "Whip"
    )

    while continuar:
      self.menu()
      escolha = input("Digite a opção desejada: ")

      match escolha:
        case "1":
          escolhertipoarma2 = self.escolhertipoarma(
              df, listatiposarmas
          )

        case "2":
          requerimentodesejado2, valorstatus2 = self.requerimentostatus(
              df, statusds
          )

        case "3":
          scalingdesejado2, scalingtipodesejado2 = self.scalingdesejado(
              df, valorscaling, scalingtipo
          )

        case "4":
          if (
              escolhertipoarma2 is not None and
              requerimentodesejado2 is not None and
              valorstatus2 is not None and
              scalingdesejado2 is not None and
              scalingtipodesejado2 is not None
          ):
            resultado = self.filtragem(
                df,
                listatiposarmas,
                escolhertipoarma2,
                statusds,
                requerimentodesejado2,
                valorstatus2,
                valorscaling,
                scalingtipo,
                scalingdesejado2,
                scalingtipodesejado2
            )
          else:
            print("Execute as opções 1, 2 e 3 antes de filtrar.")

        case "5":
          continuar = False
          print("Processo encerrado.")

        case _:
          print("Opção inválida. Tente novamente.")
```
