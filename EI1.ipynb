```python
def adicionar_notas():
  nota = int(input("Qual é a nota do aluno"))
  lnotas.append(nota)
  return lnotas
def calcular_menor_nota(lnotas):
  menor_nota = lnotas[0]
  for i in lnotas:
    if i < menor_nota:
      menor_nota = i
  return menor_nota
def calcular_maior_nota(lnotas):
  maior_nota = lnotas[0]
  for i in lnotas:
    if i > maior_nota:
      maior_nota = i
  return maior_nota
def calcular_média(lnotas):
  média_nota = 0
  for i in lnotas:
    média_nota += i
  média_nota = média_nota / len(lnotas)
  return média_nota
def calcular_aprovados(lnotas):
  aprovados = 0
  for i in lnotas:
    if i >= 6:
      aprovados += 1
  return aprovados
def calcular_reprovados(estudantes, aprovados):
  reprovados = estudantes - aprovados
  return reprovados

lnotas = []
estudantes = int(input("Quantos estudantes?"))
while estudantes > len(lnotas):
  lnotas = adicionar_notas()
print(lnotas)
menor_nota = calcular_menor_nota(lnotas)
maior_nota = calcular_maior_nota(lnotas)
média_nota = calcular_média(lnotas)
aprovados = calcular_aprovados(lnotas)
reprovados = calcular_reprovados(estudantes, aprovados)
print(f"A menor nota é {menor_nota}")
print(f"A maior nota é {maior_nota}")
print(f"A média das notas é {média_nota}")
print(f"A quantidade de aprovados é {aprovados}")
print(f"A quantidade de reprovados é {reprovados}")
```

```python
def inserir_produto():
  código_produto = int(input("Insira o código do produto"))
  nome_produto = (input("Insira o nome do produto"))
  preço_produto = float(input("Insira o preço do produto"))
  quantidade_produto = int(input("Insira a quantidade do estoque"))
  produto = {'Código': código_produto , "Nome": nome_produto, "Preço": preço_produto, "Quantidade em estoque": quantidade_produto}
  return produto
def consultar_produto(produtos):
  código_desejado = int(input("Diga o código do produto que deseja buscar."))
  for i in produtos:
    if i["Código"] == código_desejado:
      print(i)
def calcular_valor_estoque(produtos):
  valor_total = 0
  for i in produtos:
    if i["Quantidade em estoque"] > 0:
      valor_total = valor_total + (i["Preço"] * i["Quantidade em estoque"])
  return valor_total
def calcular_maior_valor(produtos):
  maior_valor = 0
  for i in produtos:
    if i["Preço"] > maior_valor:
      maior_valor = i["Preço"]
  print(f"O maior valor unitário é {maior_valor}")




produtos = []
produtos.append(inserir_produto())
print(produtos)
consultar_produto(produtos)
valor_total = calcular_valor_estoque(produtos)
print(valor_total)
calcular_maior_valor(produtos)
```
