```python
def adicionar_notas():
  nota = int(input("Qual é a nota do aluno"))
  lnotas.append(nota)
  return lnotas
def calcular_menor_nota(lnotas):
  menor_nota = lnotas[0]
  for i in lnotas:
    if i < menor_nota:
      menor_nota = i
  return menor_nota
def calcular_maior_nota(lnotas):
  maior_nota = lnotas[0]
  for i in lnotas:
    if i > maior_nota:
      maior_nota = i
  return maior_nota
def calcular_média(lnotas):
  média_nota = 0
  for i in lnotas:
    média_nota += i
  média_nota = média_nota / len(lnotas)
  return média_nota
def calcular_aprovados(lnotas):
  aprovados = 0
  for i in lnotas:
    if i >= 6:
      aprovados += 1
  return aprovados
def calcular_reprovados(estudantes, aprovados):
  reprovados = estudantes - aprovados
  return reprovados

lnotas = []
estudantes = int(input("Quantos estudantes?"))
while estudantes > len(lnotas):
  lnotas = adicionar_notas()
print(lnotas)
menor_nota = calcular_menor_nota(lnotas)
maior_nota = calcular_maior_nota(lnotas)
média_nota = calcular_média(lnotas)
aprovados = calcular_aprovados(lnotas)
reprovados = calcular_reprovados(estudantes, aprovados)
print(f"A menor nota é {menor_nota}")
print(f"A maior nota é {maior_nota}")
print(f"A média das notas é {média_nota}")
print(f"A quantidade de aprovados é {aprovados}")
print(f"A quantidade de reprovados é {reprovados}")
```

```python
def Adquirir_Valores(lista_números):
  continuar = 1
  while continuar != 0:
    número = int(input("Digite um número(digite 0 caso deseje parar). "))
    if número != 0:
      lista_números.append(número)
    else:
      continuar = 0
  return lista_números
def Quantidade_Valores(lista_números):
  Quantidade_de_Valores = 0
  for i in lista_números:
    Quantidade_de_Valores += 1
  print(f"A quantidade de valores são {Quantidade_de_Valores}. ")
def Média_Valores(lista_números):
  média = 0
  for i in lista_números:
    média += i
  média = média / len(lista_números)
  print(f"A média dos valores é {média}. ")
def Mediana_Valores(lista_números):
  Valores_Ordenados = lista_números.sort()
  if len(lista_números) % 2 == 1:
    mediana_posição = int((len(lista_números) + 1)/2)
    mediana = lista_números[mediana_posição-1]
    print(f"A mediana da lista é: {mediana}. ")
  else:
    mediana1 = int(len(lista_números)/2)
    mediana2 = lista_números[mediana1]
    mediana1 = lista_números[mediana1-1]
    mediana = (mediana2 + mediana1)/2
    print(f"A mediana da lista é: {mediana}. ")
def Maior_Valor_Valores(lista_números):
  Maior_Valor = lista_números[0]
  for i in lista_números:
    if i > Maior_Valor:
      Maior_Valor = i
  print(f"O maior valor da lista é o {Maior_Valor}. ")
def Menor_Valor_Valores(lista_números):
  Menor_Valor = lista_números[0]
  for i in lista_números:
    if i < Menor_Valor:
      Menor_Valor = i
  print(f"O menor valor da lista é o {Menor_Valor}. ")
def Pares_Valores(lista_números):
  Quantidade_Pares = 0
  for i in lista_números:
    if i % 2 == 0:
      Quantidade_Pares += 1
  print(f"Nessa lista há {Quantidade_Pares} números pares.")
def Impares_Valores(lista_números):
  Quantidade_Impares = 0
  for i in lista_números:
    if i % 2 == 1:
      Quantidade_Impares += 1
  print(f"Nessa lista há {Quantidade_Impares} números impares.")


lista_números = []
lista_números = Adquirir_Valores(lista_números)
Quantidade_Valores(lista_números)
Média_Valores(lista_números)
Mediana_Valores(lista_números)
Maior_Valor_Valores(lista_números)
Menor_Valor_Valores(lista_números)
Pares_Valores(lista_números)
Impares_Valores(lista_números)


```
