Escreva um programa que receba do usuário os valores de um conjunto de características de um personagem inventado por você.
Leia os valores e grave um arquivo para posterior manipulação.
Escolha o formato de gravação.

O arquivo não precisa ser alterado, apenas sobreescrito.
Crie uma função para ler e apresentar (print) os dados gravados.
---



```python
import json
while True:
  try:
    nome = str(input("Digite um nome."))
    break
  except ValueError:
    print("Valor inválido, digite uma string.")
while True:
  try:
    altura = float(input("Digite a altura."))
    altura = round(altura, 2)
    break
  except ValueError:
    print("Valor inválido, digite um float.")
while True:
  try:
    nivel = int(input("Digite um nível."))
    break
  except ValueError:
    print("Valor inválido, digite um inteiro.")

dados = {
    'Nome': nome,
    'Altura': altura,
    'Nível': nivel,
}
ficha = open("dados,json",  "w")
json.dump(dados, ficha)
ficha.close()
print(ficha)

```
